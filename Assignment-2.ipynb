```python
df=spark.read.option("header",True).csv("/content/cars.csv")
```

```python
df.show()
```

```python
df.printSchema()
```

```python
#Check the summary
df.describe().show()
```

```python
#select
df.select(['brand','fuel']).show()
```

```python
#add
df_revised = df.withColumn("Discounted Price", df['selling_price']-(df['selling_price']/100))
df_revised.show()
```

```python
#drop
df_final_columns = df_revised.drop('Discounted Price')
df_final_columns.show()
```

```python
# Rename the column
df_final = df_final_columns.withColumnRenamed("selling_price", "Selling_price")
df_final.show(3)
```

```python
df_final.createOrReplaceTempView("cars_data")
df_sql = spark.sql("select * from cars_data")
df_sql.show()
```

#Filter Operation

```python
df.filter("Selling_price >450000").show()
```

```python
df.filter(df['Selling_price'] >450000).select(['brand','fuel']).show()
```

```python
#multiple filter
df.filter((df['Selling_price']>300000) | (df['Selling_price'] <400000)).show()
```

```python
df.filter(df['Selling_price'] != 200000).show()
```

```python
#multiple filter
df.filter(~(df['Selling_price']>300000) & (df['Selling_price'] <350000)).show()
```

#collect()

```python
df_final = df.filter(df["Selling_price"]>8000000).collect()
```

```python
from pyspark.sql.functions import countDistinct, avg, stddev  #imported function
```

```python
#importing function
df.select(countDistinct("km_driven")).show()
```

```python
from pyspark.sql.functions import format_number  #imported formating function
```

```python
df.select(avg("km_driven").alias("Speed Driven")).show() #change name
```

```python
df.select(format_number(avg("selling_price"),2).alias("Price")).show() #rename and remove decimal places
```
